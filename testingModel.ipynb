```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
cnn = tf.keras.models.load_model(r"C:\Users\DELL\OneDrive\Desktop\fruit-vege\fruit-vegitable-image-classification\trained-model.h5")
```

```python
#load the test image
import cv2
image_path = r"C:\Users\DELL\OneDrive\Desktop\fruit-vege\fruit-vegitable-image-classification\data\test\cabbage\Image_10.jpg"
img = cv2.imread(image_path)
plt.imshow(img)
plt.title('Test image')
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
image = tf.keras.preprocessing.image.load_img(image_path, target_size=(64,64))
input_arr = tf.keras.preprocessing.image.img_to_array(image)
input_arr = np.array([input_arr])
predections = cnn.predict(input_arr)
```

```python
print(predections)
```

```python
test_dir = r"C:\Users\DELL\OneDrive\Desktop\fruit-vege\fruit-vegitable-image-classification\data\test"
test_ds = tf.keras.preprocessing.image_dataset_from_directory(
    test_dir,
    labels='inferred',
    label_mode='categorical',
    class_names=None,
    color_mode='rgb',
    batch_size=32,
    image_size=(64,64),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation='bilinear',
    follow_links=False,
    crop_to_aspect_ratio=False
)
```

```python
print(test_ds.class_names)
```
